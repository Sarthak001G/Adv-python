```python
import json

def load_data(filename):
    # Load JSON data from file
    with open(filename, "r") as file:
        return json.load(file)

def find_people_you_may_know(user_id, data):
    user_friends = {}

    # Build a mapping: user_id -> set of friends
    # FIX: used .get() to avoid KeyError if "users" or "friends" is missing
    for user in data.get("users", []):
        user_friends[user["id"]] = set(user.get("friends", []))

    # If the given user does not exist, return empty list
    if user_id not in user_friends:
        return []

    # Direct friends of the given user
    direct_friends = user_friends[user_id]
    suggestions = {}  # Stores suggested_user_id -> mutual_friend_count

    # Loop through each direct friend
    for friend in direct_friends:

        # FIX: Skip friend IDs that do NOT exist in user_friends
        # This prevents KeyError (e.g., friend ID = 3 in your JSON)
        if friend not in user_friends:
            continue

        # Loop through friends of the friend
        for mutual in user_friends[friend]:

            # Exclude:
            # 1. The user themselves
            # 2. Users who are already direct friends
            if mutual != user_id and mutual not in direct_friends:
                # Count how many mutual friends each suggested user has
                suggestions[mutual] = suggestions.get(mutual, 0) + 1

    # Sort suggestions by number of mutual friends (highest first)
    sorted_suggestions = sorted(
        suggestions.items(),
        key=lambda x: x[1],   # x[1] = mutual friend count
        reverse=True          # descending order
    )

    # Return only the user IDs (not the counts)
    return [uid for uid, _ in sorted_suggestions]

# -------------------------------
# Main execution
# -------------------------------

data = load_data("cleaned_codebook_data.json")
user_id = 1  # Example: Amit

recommendations = find_people_you_may_know(user_id, data)
print(f"People You May Know for User {user_id}: {recommendations}")
```
