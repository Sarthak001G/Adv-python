```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Load data
data = pd.read_csv('iris.csv')   # Replace with your actual CSV file path

# Split into features and label
X = data.iloc[:, :-1]   # All columns except last
y = data.iloc[:, -1]    # Last column (target)
X = data.drop(columns=["Id", "Species"])
y = data["Species"]


# Train model
model = RandomForestClassifier()
model.fit(X, y)

# Predict (on same data for demo)
predictions = model.predict([[5.1,3.5,1.4,0.2]])
print(predictions)
```

```python
datatest=pd.read_excel("iris_test.xlsx")
x_test=datatest.iloc[:,:-1]
predictions = model.predict(x_test.values)
print(predictions)
```

```python
X_train = data.drop(columns=["Id", "Species"])
y_train = data["Species"]
model.fit(X_train, y_train)
```

```python
datatest = pd.read_excel("iris_test.xlsx")

# Make sure the test file has the same feature columns
x_test = datatest[X_train.columns]   # IMPORTANT: same order + same names

predictions = model.predict(x_test)  # don't use .values
print(predictions)
```
